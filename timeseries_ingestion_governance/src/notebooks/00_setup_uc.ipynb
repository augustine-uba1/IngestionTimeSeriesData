```python
dbutils.widgets.text("catalog_name", "main")
dbutils.widgets.text("raw_schema", "iot_raw")
dbutils.widgets.text("bronze_schema", "iot_bronze")

catalog = dbutils.widgets.get("catalog_name")
raw_schema = dbutils.widgets.get("raw_schema")
bronze_schema = dbutils.widgets.get("bronze_schema")

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{raw_schema}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{bronze_schema}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.iot_silver")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.iot_gold")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.iot_audit")

# Optional: schema tags (good governance signal)
spark.sql(f"ALTER SCHEMA {catalog}.{raw_schema} SET TAGS ('layer'='raw','domain'='iot')")
spark.sql(f"ALTER SCHEMA {catalog}.{bronze_schema} SET TAGS ('layer'='bronze','domain'='iot')")
spark.sql(f"ALTER SCHEMA {catalog}.iot_silver SET TAGS ('layer'='silver','domain'='iot')")
spark.sql(f"ALTER SCHEMA {catalog}.iot_gold SET TAGS ('layer'='gold','domain'='iot')")
spark.sql(f"ALTER SCHEMA {catalog}.iot_audit SET TAGS ('layer'='audit','domain'='iot')")
```
